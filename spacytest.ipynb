# Define cosinus similarity

```python
import numpy as np
```

```python
def compute_cosine_similarity(u: np.ndarray, v: np.ndarray) -> float:
    """Compute the cosine similarity between two vectors"""

    return (u @ v) / (np.linalg.norm(u) * np.linalg.norm(v))
```

# Find words similar to "dog"

```python
import numpy as np

dog_vector = nlp("dog").vector
vectors = nlp.vocab.vectors

scores = []

for word_id in vectors.keys():
    word = nlp.vocab.strings[word_id]

    if not word.isalpha() or word.lower() == "dog":
        continue

    word_vector = vectors[word_id]

    denominator = (
        np.linalg.norm(dog_vector)
        * np.linalg.norm(word_vector)
    )

    if denominator == 0:
        continue

    similarity = np.dot(dog_vector, word_vector) / denominator

    scores.append((word, float(similarity)))

scores.sort(key=lambda x: x[1], reverse=True)

for word, score in scores[:20]:
    print(f"{word:20s} {score:.4f}")
```

# One useful experiment: Try comparing "dog", "cat", "car", and "database" using:

```python
words = ["dog", "cat", "car", "database"]

for word1 in words:
    for word2 in words:
        similarity = nlp(word1).similarity(nlp(word2))
        print(f"{word1:10s} <-> {word2:10s}: {similarity:.4f}")
```
